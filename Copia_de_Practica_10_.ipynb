**¿Como trabajar con datos dipersos?**


SciPy tiene un modulo scipy.sparse que proporciona funciones para trabajar con datos dispersos. Principalmente utilizamos dos tipos de matriz dispersas: CSC-Columna dispersa comprimida. Para una aritmetica eficiente y un rapido segmentado de columnas. CSR-Fila dispersa comprimida, para un corte rapido de filas, productos matriz-vector mas rapido.
Matriz de RSE.

Podemos crear una matriz CSR pasando un array a La funcion scipy.sparse.csr_matrix().

Ejemplo: Crea un matriz CSR a partir de un arreglo:



```python
import numpy as np
from scipy.sparse import csr_matrix

arreglo = np.array([0,0,0,0,0,1,1,0,2])
print(csr_matrix (arreglo))
```

Metodos de matrices dispersas.

Visualizacion de datos almacenados (no los elementos cero) con la propiedad data:
Ejemplo

```python
import numpy as np
from scipy.sparse import csr_matrix
arreglo = np.array([[0,0,0], [0,0,1], [1,0,2]])
print(csr_matrix(arreglo).data)
```

Para contar los elementros distitos de cero ultilizamos el metodo de cournt_nonzero(). Ejemplo.

```python
import numpy as np
from scipy.sparse import csr_matrix
arreglo = np.array([[0,0,0], [0,0,1], [1,0,2]])
print(csr_matrix(arreglo).count_nonzero)
```

Para eliminar entradas de un matix ultilizamos eliminate_zero

```python
import numpy as np
from scipy.sparse import csr_matrix
arreglo = np.array([[0,0,0], [0,0,1], [1,0,2]])
matriz =csr_matrix(arreglo)
matriz.sum_duplicates()
print(matriz)
```

convertir un CSR a CSC con el metodotocsc() ejemplo:

```python
import numpy as np
from scipy.sparse import csc_matrix

arreglo = np.array([[0,0,0],[0,0,1],[1,0,2]])
nuevo = csr_matrix(arreglo).tocsc()
print(nuevo)
```
